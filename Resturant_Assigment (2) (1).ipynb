```python
### ==============================
# Restaurant Order and Billing System
# ==============================

print("===================================")
print(" Welcome To Restaurant ")
print("===================================")

# Variable
total = 0

while True:

    print("\n========== MENU ==========")
    print("1. Burger      Rs.500")
    print("2. Pizza       Rs.1000")
    print("3. Biryani     Rs.350")
    print("4. Fries       Rs.250")
    print("5. Exit")

    choice = int(input("Enter Your Choice: "))

    if choice == 1:
        item = "Burger"
        price = 500

    elif choice == 2:
        item = "Pizza"
        price = 1000

    elif choice == 3:
        item = "Biryani"
        price = 350

    elif choice == 4:
        item = "Fries"
        price = 250

    elif choice == 5:
        break

    else:
        print("Invalid Choice!")
        continue

    # Quantity Input
    qty = int(input("Enter Quantity: "))

    # Bill Calculation
    bill = price * qty

    # Total Bill
    total = total + bill

    print(item, "Added Successfully.")
    print("Current Total = Rs.", total)

    # Ask for More Order
    more = input("Do you want to order more? (yes/no): ")

    if more.lower() != "yes":
        break

# Tax
tax = total * 0.05

# Service Charge
service = total * 0.10

# Subtotal
subtotal = total + tax + service

# Discount
if subtotal > 2000:
    discount = subtotal * 0.10
else:
    discount = 0

# Final Bill
final_bill = subtotal - discount

# Final Output
print("\n========== FINAL BILL ==========")
print("Food Bill      : Rs.", total)
print("Tax (5%)       : Rs.", tax)
print("Service (10%)  : Rs.", service)
print("Discount       : Rs.", discount)
print("--------------------------------")
print("Final Amount   : Rs.", final_bill)
print("================================")
print("Thank You! Visit Again.")
```
